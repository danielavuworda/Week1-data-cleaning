HANDLING MISSING VALUES(COLUMN BY COLUMN, WITH JUSTIFICATION)
Since there is no "correct" way to handle missing data, the right choice depends on the column and what the data is used for.

```python
#Removing records without applicat names
df = df.dropna(subset=['Applicant Name'])
```

```python
df = df.dropna(subset=['Email'])
```

```python
#Filling missing university information
df['University'] = df['University'].fillna('Not Specified')
```

```python
#Filling missing application statuses
df['Status'] = df['Status'].fillna('Under Review')
```

```python
#Rechecking the missing values. This will review the results to see which missing values remain
df.isnull().sum()
```

```python
#Removing missing values for Phone numbers and Application date
df=df.dropna(subset=['Phone'])
```

```python
#Removing missing values for Application date
df=df.dropna(subset=['Application Date'])
```

```python
df.isnull().sum()
```

## Handling Missing Values

Missing values were examined separately for each column.

1. **Applicant Name:** Records without applicant names were removed because the missing identifying information limited their usefulness for applicant follow-up.
2. **Email:** Missing email addresses were handled according to the selected retention policy, considering the importance of email for communication.
3. **University:** Missing values were replaced with "Not Specified" to retain otherwise useful applicant records.
4. **Status:** Missing values were replaced with "Under Review", following the task guide's suggested default. This assumption will be documented when interpreting the results.

The missing-value counts were checked again after cleaning to assess the effectiveness of these actions.


```python
# Check exact duplicates first
df.duplicated().sum()
```

```python
#Removing exact duplicates
df = df.drop_duplicates()
```

```python
df['Email'] = df['Email'].str.strip().str.lower()
```

```python
#checking for duplicate Email addresses
df[df.duplicated(subset=['Email'], keep=False)].sort_values('Email')
```

```python
#Remove duplicate applications by Email
df = df.drop_duplicates(subset=['Email'], keep='first')
```

```python
#Verifying the results
print("Exact duplicate rows:", df.duplicated().sum())

print(
    "Duplicate email addresses:",
    df['Email'].dropna().duplicated().sum()
)
```

```python
#Removing unnecessary space
text_columns = ['Applicant Name','Email','Phone','Domain Applied','University','Status']
```

```python
#Removing leading and trailing spaces
for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x)
```

```python
#standardize applicant names
df['Applicant Name'] = df['Applicant Name'].str.title()
```

```python
#Standardize the five domain categories
print(df['Domain Applied'].value_counts(dropna=False))
```

```python
#Standardizing application status
df['Status'] = df['Status'].str.strip().str.title()
print(df['Status'].value_counts(dropna=False))
```

```python
#Correcting Date Formatting
#Convert application dates
df['Application Date'].head(10)
```

```python
df['Application Date'] = pd.to_datetime(df['Application Date'],errors='coerce',dayfirst=True)
```

```python
df['Application Date'].dtype
```

```python
#Checking missing new dates
df['Application Date'].isnull().sum()
```

```python
#Cleaning Phone number
df['Phone'] = (df['Phone'].astype('string').str.replace(r'[\s-]+', '', regex=True))
```

```python
#verifying changes
df.info()
```

```python
#Performing the final data quality checks
print("Final dataset shape:", df.shape)
#Ckecking if there are still som,e missing values
print(df.isnull().sum())
#Checking for if there are still some duplicates
print("Exact duplicates:", df.duplicated().sum())
#Checking for duplicate Emails
print("Duplicate emails:",df['Email'].dropna().duplicated().sum())
#Checking for domain categories
print(df['Domain Applied'].value_counts(dropna=False))
print(df['Domain Applied'].unique())
#Checking the final data types
df.info()
```
